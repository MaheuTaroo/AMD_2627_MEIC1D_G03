```python
BANNER = \
"""
# ----------------------------------------------------------------------
# PTS | Paulo Trigo Silva
# mLDm | MoP
# v01
# ----------------------------------------------------------------------
"""

# notebook evidence cell | version to complete by the student
#
# Instructions:
# 1. copy this code into one notebook cell
# 2. complete the TODO values
# 3. execute the cell
# 4. check that the evidence file was created in 04_report/

from pathlib import Path
from datetime import datetime
import importlib
import platform
import sys


FOLDER_NAME_WORKSPACE = "mLDm_MoP_workspace"
FOLDER_NAME_MOP = "MoP_01_install_and_test"

# TODO: complete the expected name of the virtual-environment folder
# hint: it is the folder created directly under mLDm_MoP_workspace/
FOLDER_NAME_VENV = "__venv"

# TODO: complete the report folder name used in each MoP.
FOLDER_NAME_REPORT = "04_report"

# TODO: complete the output file name to be generated by this notebook cell.
OUTPUT_FILE_NAME = "01_my_notebook_check.txt"

PACKAGE_LIST = [
   ( "numpy", "numpy" ),
   ( "pandas", "pandas" ),
   ( "duckdb", "duckdb" ),
   ( "psycopg", "psycopg" ),
   ( "sklearn", "scikit-learn" ),
   ( "matplotlib", "matplotlib" ),
   ( "plotly", "plotly" ),
   ( "jupyterlab", "jupyterlab"), 
   ( "ipykernel", "ipykernel"),
   ( "ipywidgets", "ipywidgets")
   # TODO: add the missing Jupyter-related package/module pairs.
   # Use the same format:
   # ( "module_name", "package-name" ),
]


def find_workspace_path():
   current_path = Path.cwd().resolve()

   for path in [ current_path, *current_path.parents ]:
      if path.name == FOLDER_NAME_WORKSPACE:
         return path

      workspace_candidate_path = path / FOLDER_NAME_WORKSPACE
      if workspace_candidate_path.is_dir():
         return workspace_candidate_path

   raise FileNotFoundError(
      f"Could not find the workspace folder: {FOLDER_NAME_WORKSPACE}" )


def find_mop_path( workspace_path ):
   current_path = Path.cwd().resolve()

   for path in [ current_path, *current_path.parents ]:
      if path.name == FOLDER_NAME_MOP:
         return path

   mop_candidate_path = workspace_path / FOLDER_NAME_MOP
   if mop_candidate_path.is_dir():
      return mop_candidate_path

   raise FileNotFoundError(
      f"Could not find the MoP folder: {FOLDER_NAME_MOP}" )


def detect_virtual_environment( workspace_path ):
   executable_path = Path( sys.executable ).resolve()

   # TODO: build the expected virtual-environment path.
   # It should combine workspace_path with FOLDER_NAME_VENV.
   expected_venv_path = workspace_path / FOLDER_NAME_VENV

   return expected_venv_path in executable_path.parents


def check_required_packages():
   result_list = []
   failed_import_list = []

   for module_name, package_name in PACKAGE_LIST:
      try:
         module = importlib.import_module( module_name )
         version = getattr( module, "__version__", "version-not-available" )
         result_list.append( f"{package_name}: OK | {version}" )
      except Exception as error:
         failed_import_list.append( package_name )
         result_list.append( f"{package_name}: ERROR | {error}" )

   return result_list, failed_import_list


workspace_path = find_workspace_path()
mop_path = find_mop_path( workspace_path )
report_path = mop_path / FOLDER_NAME_REPORT
output_file_path = report_path / OUTPUT_FILE_NAME

kernel_uses_venv = detect_virtual_environment( workspace_path )
package_result_list, failed_import_list = check_required_packages()

validation_lines = [
   "PTS | mLDm | notebook evidence",
   f"timestamp: {datetime.now().isoformat( timespec = 'seconds' )}",
   f"platform: {platform.platform()}",
   f"workspace_path: {workspace_path}",
   f"mop_path: {mop_path}",
   f"python_executable: {sys.executable}",

   # TODO: complete this line so that it reports yes/no.
   # hint: use the variable kernel_uses_venv.
   f"kernel_uses_virtual_environment: {"yes" if kernel_uses_venv else "no"}",

   "",
   "package_check:"
]

validation_lines.extend( package_result_list )

validation_lines.extend(
   [
      "",
      f"required_packages_imported: {'yes' if len( failed_import_list ) == 0 else 'no'}",
      f"failed_imports: {', '.join( failed_import_list ) if failed_import_list else 'none'}",
      "widget_displayed: yes/no",
      "problems_found: <describe, if any>"
   ]
)

validation_text = "\n".join( validation_lines )

print( validation_text )

report_path.mkdir( parents = True, exist_ok = True )
output_file_path.write_text( validation_text, encoding = "utf-8" )

print()
print( f"validation file written to: {output_file_path}" )
```
